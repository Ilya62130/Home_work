# Диапазон дат

Функция возвращает список дат от начальной до конечной включительно. Даты передаются в формате `YYYY-MM-DD`.

```python
import datetime as dt


def date_range(start_date: str, end_date: str) -> list[dt.datetime]:
    date_format = "%Y-%m-%d"
    start = dt.datetime.strptime(start_date, date_format)
    end = dt.datetime.strptime(end_date, date_format)

    return [
        start + dt.timedelta(days=day)
        for day in range((end - start).days + 1)
    ]


print(date_range("2026-09-01", "2026-11-01"))
```

Задание 1

```python
import requests
```

```python
url = "https://jsonplaceholder.typicode.com/posts"
responce = requests.get(
    url,
    headers={'Accept': 'aplication/json'},
    timeout=10
)
```

```python
posts = responce.json()
```

```python
for post in posts:
    print(f"Пользователь: {post['userId']}")
    print(f"Заголовок: {post['title']}")
    print(f"Текст: {post['body']}")
    print()
```

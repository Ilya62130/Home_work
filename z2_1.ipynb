Задание 1.
Функция sum_distance(from, to) суммирует все числа от значения from до величины to включительно

### Оптимизация

Формула суммы арифметической прогрессии выполняется за постоянное время **O(1)** и не перебирает весь диапазон.

```python
def sum_distance(from_: int, to_: int) -> int:
    """Суммирует целые числа между границами включительно."""
    start, end = sorted((from_, to_))
    count = end - start + 1
    return (start + end) * count // 2


print(sum_distance(1, 5))
print(sum_distance(5, 1))
```

Функция sum_distance(from, to) суммирует все числа от значения from до величины to включительно

```python
def sum_distance(from_:int, to_:int) -> int:
    if from_ > to_:
        from_, to_ = to_, from_

    return sum(range(from_, to_ + 1))


print(sum_distance(1, 5))
print(sum_distance(5, 1))
```

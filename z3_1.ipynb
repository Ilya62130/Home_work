Переведите содержимое файла purchase_log.txt в словарь purchases вида:
 {'1840e0b9d4': 'Продукты', ...}

Пример работы программы при выводе первых двух элементов словаря purchases:

1840e0b9d4 ‘Продукты‘

4e4f90fcfb  ‘Электроника‘

```python
import json


def load_purchases(file_path: str) -> dict[str, str]:
    purchases: dict[str, str] = {}

    with open(file_path, "r", encoding="utf-8") as file:
        next(file)  # пропускаем строку-заголовок

        for line in file:
            purchase = json.loads(line)
            purchases[purchase["user_id"]] = purchase["category"]

    return purchases


```

```python
purchases = load_purchases("purchase_log.txt")

for user_id, category in list(purchases.items())[:10]:
    print(user_id, category)
```
